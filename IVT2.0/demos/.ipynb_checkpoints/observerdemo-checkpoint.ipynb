```julia
using Pkg
Pkg.activate("../IVTmodel")
Pkg.instantiate()
include("../modules/IVTmodel.jl")
#Generates settings for parameters used
fittingmodel = setupmodel_IVT2()
#Imports calibration data from csv files
modeldata = importakamadata()
#Imports PPiase validation data
PPiasedata = importakamaPPiasedata();
akamafittedparametersmatrix = Matrix(CSV.read("../outputs/fittedparameters.csv", DataFrame,header=false))
akamafittedparameterslist = reshape(akamafittedparametersmatrix,(size(akamafittedparametersmatrix)[1],))
akamafittedcovariancematrix = Matrix(CSV.read("../outputs/covariancematrix.csv", DataFrame,header=false))
akamafittedparams = fullparameterset(fittingmodel,akamafittedparameterslist)
printresidual(fittingmodel,modeldata,akamafittedparameterslist)
```

```julia
ts = LinRange(0,3.2,1000)
phvalues = [discreteRNA(t) for t in ts]
plot(ts,phvalues)
```

```julia
plt1 = plot(ts,phvalues,label = "Sensor Data",ylabel = "pH",xlabel = "Time (h)")
plt2 = scatter([0],[0],label = "",legend = :bottomright,ylabel = "RNA concentration (μM)",xlabel = "Time (h)")
inputs = (T7RNAP = 240e-9, ATP = 8e-3,UTP = 8e-3,CTP = 8e-3,GTP = 8e-3, Mg = 26e-3, Buffer = 0.040, DNA = 19.442, final_time = 2.2)
PPiase = 0.02
Cap = 4e-3
stoich = (531,309,692,588)

sol = runobservedbatch(akamafittedparams, inputs,-0.0,discreteRNA,PPiase = PPiase, stoich = stoich)
plot!(plt1,sol.t,ph(sol,0),linewidth = 3, label = "Model - No observer")
plot!(plt2,sol.t,utp(sol,0),linewidth = 3, label = "Model - No observer")



# sol = runobservedbatch(akamafittedparams, inputs,-0.005,phmeasure,discretepH,PPiase = 0.01, stoich = (0,1.65,3.25,5))
# plot!(plt1,sol.t,ph(sol,0),linewidth = 3, label = "Model - Weak tuning")
# plot!(plt2,sol.t,utp(sol,0),linewidth = 3, label = "Model - Weak tuning")
# sol = runobservedbatch(akamafittedparams, inputs,-0.1,phmeasure,discretepH,PPiase = 0.01, stoich = (0,1.65,3.25,5))
# plot!(plt1,sol.t,ph(sol,0),linewidth = 3, label = "Model - Strong Tuning")
# plot!(plt2,sol.t,utp(sol,0),linewidth = 3, label = "Model - Strong Tuning")
# fullplot = plot(plt1,plt2, size = (800,400),bottommargin = 5mm,leftmargin = 5mm,dpi = 2000)
# savefig(fullplot,"../figures/pHobserver.png")
# fullplot
```
